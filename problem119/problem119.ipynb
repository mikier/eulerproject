```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python

def sum_of_digits(n):
    return sum([int(x) for x in str(n)])

RANGE_X=100
RANGE_Y=30
df = pd.DataFrame([(x,y, x**y, sum_of_digits(x**y), x==sum_of_digits(x**y)) for x in range(2,RANGE_X) for y in range(2,RANGE_Y)],columns=['x','y','x**y','sumx**y','is_equal'])

```

```python
r=df[df['is_equal']].sort_values('x**y')
```
