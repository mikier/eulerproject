```python
import numpy as np
import pandas as pd
```

```python
def factor(n):
    fact = []
    for i in range(n+1):
        if i == np.gcd(i,n):
            fact.append(i)
    return fact , len(fact)==2


def all_prime_until(N):
    prime_list=[]
    prime_list_until = [] 
    for i in range(N):
        if factor(i)[1]:
            prime_list.append(i)
        prime_list_until.append(prime_list.copy())
    return prime_list_until

N = 10001
ALL_PRIME_UNTIL_N = all_prime_until(N)
```

```python
def factorize(n):
    factors = []
    if n==1:
        return []
    for i in ALL_PRIME_UNTIL_N[n]:
        if n%i==0:
            factors.append(i)
            return [i] + factorize(n//i)
    
def unique_factor(n):
    factors_list = factorize(n)
    factor_set = set(factors_list)
    unique_factors = []
    for i in factor_set:
        unique_factors.append((i, factors_list.count(i)))
    return unique_factors
```
